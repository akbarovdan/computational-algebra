2. A mountain climber uses the model `T = 20 - 10h` to estimate the temperature T (in C) at elevation h (in kilometers, km).
- Make a table that gives the temperature for each 1-km change in elevation, from elevation 0 km to elevation 5 km. How does temperature change as elevation ­increases?

```python
for h in range(0, 6):
    temp = 20 - 10 * h
    print(f'{h}km = {temp}c')
```

```python
import pandas as pd

df = pd.DataFrame({'Elevation (km)': range(0, 6)})

df['Temperature (°C)'] = 20 - 10 * df['Elevation (km)']

print(df)
```

- If the temperature is 5°C, what is the elevation?

```python
# if temp = 20 - 10 * h then h = (20 - temp) / 10
temp = 5
print(f'h = {(20 - temp) / 10}')
```

3. Henry’s car used 10.5 gal to drive 230 mi. Find its gas mileage.

```python
gal = 10.5
mi = 230

mileage = mi / gal

print(f"Gas mileage: {mileage:.2f} miles/gal")
```
