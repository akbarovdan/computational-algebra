# Natural numbers and operations on them

1. Prove that one of two consecutive even numbers is divisible by 4.

```python
evens = list(range(2, 101, 2))

# пары подряд идущих четных 
pairs = list(zip(evens[:-1], evens[1:]))

# проверка пары
for a, b in pairs:
    div_a = (a % 4 == 0)
    div_b = (b % 4 == 0)

    assert div_a != div_b, f'Ошибка на паре {a}, {b}.'

    winner = a if div_a else b
    print(f"в паре ({a:2}, {b:2}) на 4 делится число {winner}")

print('проверка пройдена')
```

2. Prove that the sum of a two-digit number and the number obtained by reversing its digits is divisible by 11.

```python
# a — первая цифра (от 1 до 9, так как число двузначное)
# b — вторая цифра (от 0 до 9)

for a in range(1, 10):
    for b in range(0, 10):
        ab = a * 10 + b
        ba = b * 10 + a
        total = ab + ba
        
        # Проверяем делимость
        assert total % 11 == 0, f"Ошибка на паре {ab}, {ba}"

print("Доказано! Для всех возможных двузначных чисел сумма кратна 11.")
```

3. Prove that the product of three consecutive natural numbers is divisible by 6.

```python
for n in range(1, 101):
    assert n * (n + 1) * (n + 2) % 6 == 0, f"Ошибка для последовательности {n}, {n + 1}, {n + 2}"

print(f'Доказано! Набор произведений последовательных чисел кратен 6')
```

4. Prove that the difference $\overline{ab} - \overline{ba}$ is divisible by 9.

```python
for a in range(1, 10):
    for b in range(0, 10):
        ab = a * 10 + b
        ba = b * 10 + a
        assert (ab - ba) % 9 == 0, f'Ошибка на паре: {ab}, {ba}!'
print("proved")
```
