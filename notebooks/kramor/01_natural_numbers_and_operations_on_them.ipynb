# Natural numbers and operations on them

1. Prove that one of two consecutive even numbers is divisible by 4.

```python
evens = list(range(2, 101, 2))

# пары подряд идущих четных 
pairs = list(zip(evens[:-1], evens[1:]))

# проверка пары
for a, b in pairs:
    div_a = (a % 4 == 0)
    div_b = (b % 4 == 0)

    assert div_a != div_b, f'Ошибка на паре {a}, {b}.'

    winner = a if div_a else b
    print(f"в паре ({a:2}, {b:2}) на 4 делится число {winner}")

print('проверка пройдена')
```
