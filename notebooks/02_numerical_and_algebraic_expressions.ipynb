### **Problem 32**

Write as a numerical expression:

* **a)** the sum of the number $25$ and the product of $16$ and $74$
* **b)** the difference between the product of $37$ and $6$ and the product of $29$ and $5$
* **c)** the product of the difference of $86$ and $17$ and their sum
* **d)** the quotient of the number $98$ and the sum of $9$ and $5$

```python
print(25 + (16 * 74))
print((37 * 6) - (29 * 5))
print((86 - 17) * (86 + 17))
print(98 / (9 + 5))
```

### **Problem 33**

Determine which of the numerical expressions are undefined (have no meaning):

1. $(2.5 - 5.5 \div 2.2) \div (4.8 - 0.3 - 1.3)$

2. $(2.5 - 5.5 \div 1.1) \div \left(4.8 - 0.4 \cdot 1\frac{1}{5}\right)$

3. $\frac{5.5 \div 2.2 - 2.5}{4.8 - 0.3 \cdot 1.3}$

4. $(2.8 - 5.5 \div 2.2) \cdot (4.8 + 0.3 \cdot (-1.6))$

```python
def check_expression(exp_lambda):
    try:
        val = exp_lambda()
        return f"defined, результат = {val}"
    except ZeroDivisionError:
        return "Undefined: деление на ноль."

# Передаем выражения
print(f"a) {check_expression(lambda: (2.5 - 5.5 / 2.2) / (4.8 - 0.3 - 1.3))}")
print(f"b) {check_expression(lambda: (2.5 - 5.5 / 1.1) / (4.8 - 0.4 * (1 + 1/5)))}")
print(f"c) {check_expression(lambda: (5.5 / 2.2 - 2.5) / (4.8 - 0.3 * 1.3))}")
print(f"d) {check_expression(lambda: (2.8 - 5.5 / 2.2) * (4.8 + 0.3 * (-1.6)))}")
```
