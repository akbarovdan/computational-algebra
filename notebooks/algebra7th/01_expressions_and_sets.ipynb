### **Problem 21**

Let $L$ be the set of single-digit natural numbers. By listing its elements (using roster notation), specify the subset of $L$ consisting of:

* **a)** prime numbers
* **b)** even numbers
* **c)** odd numbers
* **d)** multiples of $9$

```python
L = set(range(1, 10))
primes = {2, 3, 5, 7, 11, 13, 17, 19}

A = {x for x in L if x in primes}
print(f'a) {A}')

B = {x for x in L if x % 2 == 0}
print(f'б) {B}')

C = {x for x in L if x % 2 !=  0}
print(f'в) {C}')

D = {x for x in L if x % 9 == 0}
print(f'г) {D}')
```

### **Problem 24**

Let $V$ be the set of unit fractions (fractions with a numerator equal to $1$). By listing its elements, specify the subset of $V$ consisting of fractions:

* **a)** with denominators $7$, $6$, $5$
* **b)** greater than $\frac{1}{10}$
* **c)** greater than $\frac{1}{16}$ and less than $\frac{1}{9}$
* **d)** greater than $\frac{2}{5}$

```python
from fractions import Fraction as fr

V = {fr(1, x) for x in range(1, 21)}

a_list = [7, 6, 5]
A = {x for x in V if x.denominator in a_list}
print(f'a) A = {A}')

B = {x for x in V if x.denominator < 10}
print(f'b) B = {B}')

C = {x for x in V if 9 < x.denominator < 16}
print(f'c) C = {C}')

D = {x for x in V if x > fr(2, 5)}
print(f'd) D = {D}')
```

### **Problem 25**

From the set of two-digit numbers, select the subset of numbers in which:

* **a)** the sum of the digits is equal to $9$
* **b)** the difference between the tens digit and the units digit is equal to $8$
* **c)** the product of the tens digit and the units digit is equal to $6$
* **d)** the product of the tens digit and the units digit is equal to the sum of the tens digit and the units digit

```python
M = set(range(10, 100))

A = {x for x in M if x // 10 + x % 10 == 9}
print(f'a) A = {A}')

B = {x for x in M if x // 10 - x % 10 == 8}
print(f'b) B = {B}')

C = {x for x in M if (x // 10) * (x % 10) == 6}
print(f'c) C = {C}')

D = {x for x in M if (x // 10) * (x % 10) == x // 10 + x % 10}
print(f'd) D = {D}')
```

### **Problem 28**

List all subsets of set $A$ if:

* **a)** $A = \{1, 5\}$
* **b)** $A = \{8\}$
* **c)** $A = \{2, 3, 4\}$

```python
import itertools

def all_subsets(some_set):
    power_set = []
    for r in range(len(some_set) + 1):
        for combo in itertools.combinations(some_set, r):
            power_set.append(list(combo))
    print(f'Set {some_set} contains the following subsets: {power_set}')

a_set = [1, 5]
b_set = [8]
c_set = [2, 3, 4]

all_subsets(a_set)
all_subsets(b_set)
all_subsets(c_set)
```

Let $A$ be the set of numbers that are multiples of $2$, $B$ be the set of numbers that are multiples of $5$, and $C = \{135, 253, 374, 470, 586, 721\}$. List all elements of set $C$ that:

### **Problem 30**

$3.6$ tons of granulated sugar were brought to a warehouse. Then $75\%$ of this amount was sent to a store and a kiosk, with $0.9$ tons more sent to the store than to the kiosk. How much granulated sugar was sent to the kiosk?

```python
# Known data
total_warehouse = 3.6          # Total tons at warehouse
percent_sent = 75 / 100        # 75%
difference = 0.9               # Store received 0.9 tons more than kiosk

# Step 1: Calculate total sugar sent to store and kiosk combined
total_sent = total_warehouse * percent_sent

# Step 2 & 3: Solve the system (Store + Kiosk = total_sent, Store = Kiosk + 0.9)
# 2 * Kiosk = total_sent - difference
kiosk_sugar = (total_sent - difference) / 2
store_sugar = kiosk_sugar + difference

print(f'Total sugar sent: {total_sent:.2f} tons')
print(f'Sugar sent to store: {store_sugar:.2f} tons')
print(f'Sugar sent to kiosk: {kiosk_sugar:.2f} tons')
```

### **Problem 31**

Find the number if it is known that:

* **a)** $2\%$ of this number is equal to $3.5$
* **b)** $70\%$ of this number is equal to $29.4$
* **c)** $140\%$ of this number is equal to $112$

```python
a = {'percent': 2, 'value': 3.5}
b = {'percent': 70, 'value': 29.4}
c = {'percent': 140, 'value': 112}

def find_number(data_dict):
    return data_dict['value'] * 100 / data_dict['percent']

print(f'а) Number is equal: {find_number(a)}')
print(f'b) Number is equal: {find_number(b)}')
print(f'c) Number is equal: {find_number(c)}')
```
